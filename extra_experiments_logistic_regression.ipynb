```python
import numpy as np
from problems.sparse_logistic_regression import Sparse_Logistic_Regression

import matplotlib.pyplot as plt
import pandas as pd

from first_order_methods import First_Order_Methods
from libsvmdata import fetch_libsvm
import pickle
```

# a1a - a9a

```python
for i in range(1, 10):
    dataset = "a%da"%i
    c = 0.001
    k = 10000
    right_lim = 11000
    print("========================")
    A, b = fetch_libsvm(dataset)
    title = "a%da: m=%d, n=%d, c=0.001"%(i, A.shape[0], A.shape[1])
    print(title)
    slr = Sparse_Logistic_Regression(A, b, c)
    print("gamma: %f"%slr.gamma)
    fom = First_Order_Methods(slr)
    print("initial function value: %f"%slr.objective(fom.x_0))
    print("------------------------")
    result_slr = fom.test_all_methods_smooth(k=k, beta=0.184, draw_performance_plot=True, dataset=title, initial_line_search=False, right_lim=right_lim)
    with open('logistic_a%da_001.pkl'%i, 'wb') as f: 
        pickle.dump(result_slr, f)
```
